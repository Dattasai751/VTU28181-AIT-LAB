```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}
def bfs_recursive(graph, queue, visited, target=None):
    if not queue:
        return False
    current = queue.popleft()
    print(current, end=" ")
    if current == target:
        return True
    for neighbor in graph[current]:
        if neighbor not in visited:
            visited.add(neighbor)
            queue.append(neighbor)
    return bfs_recursive(graph, queue, visited, target)
def bfs(graph, start, target=None):
    visited = set()
    queue = deque()
    visited.add(start)
    queue.append(start)
    print("BFS Traversal:", end=" ")
    found = bfs_recursive(graph, queue, visited, target)
    print()
    if target is not None:
        if found:
            print("Target node", target, "found.")
        else:
            print("Target node", target, "not found.")
def dfs_recursive(graph, node, visited, target=None):
    if node in visited:
        return False
    visited.add(node)
    print(node, end=" ")
    if node == target:
        return True
    for neighbor in graph[node]:
        if dfs_recursive(graph, neighbor, visited, target):
            return True
    return False
def dfs(graph, start, target=None):
    visited = set()
    print("DFS Traversal:", end=" ")
    found = dfs_recursive(graph, start, visited, target)
    print()

    if target is not None:
        if found:
            print("Target node", target, "found.")
        else:
            print("Target node", target, "not found.")
print("GRAPH SEARCH ALGORITHM")
print("\nGraph Representation:")
for node in graph:
    print(node, "->", graph[node])
print("\nBREADTH FIRST SEARCH")
bfs(graph, 'A')
print("\nDEPTH FIRST SEARCH")
dfs(graph, 'A')
print("\nBFS SEARCH FOR NODE F")
bfs(graph, 'A', 'F')
print("\nDFS SEARCH FOR NODE F")
dfs(graph, 'A', 'F')
```

```python
import heapq
n = int(input("Enter number of nodes: "))
graph = {}
for i in range(n):
    node = input(f"Enter node {i + 1}: ")
    graph[node] = []
edges = int(input("\nEnter number of edges: "))
print("\nEnter edges in the format:")
print("Source Destination Cost")
for i in range(edges):
    source, destination, cost = input(f"Edge {i + 1}: ").split()
    cost = int(cost)
    graph[source].append((destination, cost))
heuristic = {}
print("\nEnter heuristic h(n) value for each node:")
for node in graph:
    heuristic[node] = int(input(f"h({node}) = "))
start = input("\nEnter starting node: ")
goal = input("Enter goal node: ")
open_list = []
g_cost = {start: 0}
parent = {}
f_start = g_cost[start] + heuristic[start]
heapq.heappush(open_list, (f_start, start))
visited = set()
while open_list:
    f_current, current = heapq.heappop(open_list)
    if current in visited:
        continue
    visited.add(current)
    print("\nCurrent Node:", current,
          "g(n) =", g_cost[current],
          "h(n) =", heuristic[current],
          "f(n) =", f_current)
    if current == goal:
        break
    for neighbor, cost in graph[current]:
        new_g = g_cost[current] + cost
        if neighbor not in g_cost or new_g < g_cost[neighbor]:
            g_cost[neighbor] = new_g
            f_value = new_g + heuristic[neighbor]
            parent[neighbor] = current
            heapq.heappush(open_list, (f_value, neighbor))
if goal in g_cost:
    path = []
    current = goal
    while current != start:
        path.append(current)
        current = parent[current]
    path.append(start)
    path.reverse()
    print("\n..................")
    print("A* SEARCH RESULT")
    print("..................")
    print("Optimal Path:", " -> ".join(path))
    print("Path Cost:", g_cost[goal])
else:
    print("\nNo path found.")
```

```python
class Node:
    def __init__(self, name, value=None, children=None):
        self.name = name
        self.value = value
        self.children = children if children is not None else []
    def is_terminal(self):
        return len(self.children) == 0
def minimax(node, depth, is_maximizing_player):
    if depth == 0 or node.is_terminal():
        return node.value
    if is_maximizing_player:
        max_eval = float('-inf')
        for child in node.children:
            evaluation = minimax(child, depth - 1, False)
            max_eval = max(max_eval, evaluation)
        return max_eval
    else:
        min_eval = float('inf')
        for child in node.children:
            evaluation = minimax(child, depth - 1, True)
            min_eval = min(min_eval, evaluation)
        return min_eval
leaf_A = Node("A", value=3)
leaf_B = Node("B", value=5)
leaf_C = Node("C", value=2)
leaf_D = Node("D", value=9)
leaf_E = Node("E", value=12)
leaf_F = Node("F", value=5)
leaf_G = Node("G", value=23)
leaf_H = Node("H", value=23)
child_node_1 = Node("Child_1", children=[leaf_A, leaf_B])
child_node_2 = Node("Child_2", children=[leaf_C, leaf_D])
child_node_3 = Node("Child_3", children=[leaf_E, leaf_F])
child_node_4 = Node("Child_4", children=[leaf_G, leaf_H])
branch_left = Node("Branch_Left",
                   children=[child_node_1, child_node_2])
branch_right = Node("Branch_Right",
                    children=[child_node_3, child_node_4])
root = Node("Root",
            children=[branch_left, branch_right])
optimal_value = minimax(root, depth=3, is_maximizing_player=True)
print("The optimal value at the Root node is:", optimal_value)
```
